```python
'''
What is an outlier?
An outlier is a data point that significantly differs from other observations in a dataset. It can be an unusually high or low value that does not fit the general pattern of the data. Outliers can occur due to various reasons, such as measurement errors, data entry mistakes, or natural variability in the data.
Why is outlier detection important?
Outlier detection is crucial for several reasons:
1. Data Quality: Outliers can indicate errors in data collection or entry, which can affect the quality of the dataset and lead to inaccurate analysis.
2. Statistical Analysis: Outliers can skew statistical measures such as mean and standard deviation, leading
to misleading conclusions. Detecting and handling outliers can help ensure that statistical analyses are more accurate and representative of the underlying data.
3. Machine Learning: Outliers can negatively impact the performance of machine learning models, as they can influence the model's ability to learn patterns in the data. Detecting and handling outliers can improve model performance and generalization.
4. Decision Making: Outliers can provide valuable insights into unusual events or behaviors that may require special attention. Detecting outliers can help identify potential issues or opportunities in the data that may not be apparent from the overall trends.
Methods for Outlier Detection:
1. Z-Score Method: This method calculates the z-score for each data point, which measures how many standard deviations a data point is from the mean. Data points with a z-score greater than a certain threshold (commonly 3) are considered outliers.
2. Interquartile Range (IQR) Method: This method uses the IQR to identify outliers. Data points that fall below Q1 - 1.5*IQR or above Q3 + 1.5*IQR are considered outliers, where Q1 and Q3 are the first and third quartiles, respectively.
3. Visualization: Box plots, scatter plots, and histograms can help visually identify outliers in the data.
4. Machine Learning Algorithms: Some algorithms, such as Isolation Forest and Local Outlier Factor, are specifically designed for outlier detection in high-dimensional data.
Handling Outliers:
1. Removal: If outliers are determined to be errors or irrelevant to the analysis, they can be removed from the dataset.
2. Transformation: Applying transformations (e.g., log transformation) can help reduce the impact of outliers on the analysis.
3. Imputation: Outliers can be replaced with more representative values, such as the mean or median of the data.
4. Robust Models: Using machine learning models that are less sensitive to outliers, such as decision trees or random forests, can help mitigate the impact of outliers on model performance.
'''

# Using Z-Score Method to detect outliers in the 'CashbackAmount' column
from scipy import stats
import numpy as np
# Calculate z-scores for the 'CashbackAmount' column
data['z_score'] = np.abs(stats.zscore(data['CashbackAmount']))
# Define a threshold for identifying outliers
threshold = 3
# Identify outliers based on the z-score threshold
outliers = data[data['z_score'] > threshold]
print("Outliers detected using Z-Score Method:")
print(outliers[['CashbackAmount', 'z_score']])
```

```python
# Using IQR Method to detect outliers in the 'CashbackAmount' column
# Calculate Q1 and Q3
Q1 = data['CashbackAmount'].quantile(0.25)
Q3 = data['CashbackAmount'].quantile(0.75)
# Calculate IQR
IQR = Q3 - Q1
# Define lower and upper bounds for outliers
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
# Identify outliers based on the IQR method
outliers_iqr = data[(data['CashbackAmount'] < lower_bound) | (data['CashbackAmount'] > upper_bound)]
print("Outliers detected using IQR Method:")
print(outliers_iqr[['CashbackAmount']])
```
