Day 15 Categorical Encoding: Turning Categories into Turnable Signals.

```python
'''
Encoding means converting data from one format to another, typically for the purpose of standardization, speed, or compression. Decoding is the reverse process, converting encoded data back to its original format.
Categorical encoding is the process of converting categorical data (data that can take on a limited, fixed number of possible values) into a numerical format that can be used in machine learning algorithms.
Types of Categorical Encoding:
1. One-Hot Encoding: Converts each category value into a new binary column (0 or 1).
2. Label Encoding: Assigns each unique category value an integer label.
3. Ordinal Encoding: Similar to label encoding but maintains the order of categories.
'''

import pandas as pd
from sklearn.preprocessing import OneHotEncoder, LabelEncoder, OrdinalEncoder

data  = pd.read_excel(r"C:\Users\MUSKAAN\Documents\M.E Folder\ML&AI\Machine Learning Journey\ml-from-basics-to-beyond\Phase_2_Data_Handling\E Commerce Dataset.xlsx", sheet_name="E Comm")
data.head(5)
```

```python
data.info()
```

```python
data.isnull().sum()

# We will be using MaritalStatus,Gender columns for encoding using pandas get_dummies.
# One-Hot Encoding
en_data_1 = ["Gender","MaritalStatus"]
pd.get_dummies(data[en_data_1], drop_first=True).head(5)
```

```python
# To use OneHotEncoder from sklearn, we need to reshape the data and fit the encoder.
# One-Hot Encoding using sklearn
ohe = OneHotEncoder(drop='first')

# Fit and transform the data means we are applying the encoding to the specified columns and converting them into a format suitable for machine learning models.
ohe_data = ohe.fit_transform(data[en_data_1])

ohe_data.toarray()[:5]  # Convert sparse matrix to array and display first 5 rows

# To convert the encoded data back to a DataFrame with appropriate column names:
ohe_df = pd.DataFrame(ohe_data.toarray(), columns=ohe.get_feature_names_out(en_data_1))
ohe_df.head(5)
```
