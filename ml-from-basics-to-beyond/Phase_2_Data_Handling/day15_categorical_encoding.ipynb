Day 15 Categorical Encoding: Turning Categories into Turnable Signals.

```python
'''
Encoding means converting data from one format to another, typically for the purpose of standardization, speed, or compression. Decoding is the reverse process, converting encoded data back to its original format.
Categorical encoding is the process of converting categorical data (data that can take on a limited, fixed number of possible values) into a numerical format that can be used in machine learning algorithms.
Types of Categorical Encoding:
1. One-Hot Encoding: Converts each category value into a new binary column (0 or 1).
2. Label Encoding: Assigns each unique category value an integer label.
3. Ordinal Encoding: Similar to label encoding but maintains the order of categories.
'''

import pandas as pd
from sklearn.preprocessing import OneHotEncoder, LabelEncoder, OrdinalEncoder

data  = pd.read_excel(r"C:\Users\MUSKAAN\Documents\M.E Folder\ML&AI\Machine Learning Journey\ml-from-basics-to-beyond\Phase_2_Data_Handling\E Commerce Dataset.xlsx", sheet_name="E Comm")
data.head(5)
```

```python
data.info()
```

```python
data.isnull().sum()

# We will be using MaritalStatus,Gender columns for encoding using pandas get_dummies.
# One-Hot Encoding
en_data_1 = ["Gender","MaritalStatus"]
pd.get_dummies(data[en_data_1], drop_first=True).head(5)
```

```python
# To use OneHotEncoder from sklearn, we need to reshape the data and fit the encoder.
# One-Hot Encoding using sklearn
ohe = OneHotEncoder(drop='first')

# Fit and transform the data means we are applying the encoding to the specified columns and converting them into a format suitable for machine learning models.
ohe_data = ohe.fit_transform(data[en_data_1])

ohe_data.toarray()[:5]  # Convert sparse matrix to array and display first 5 rows

# To convert the encoded data back to a DataFrame with appropriate column names:
ohe_df = pd.DataFrame(ohe_data.toarray(), columns=ohe.get_feature_names_out(en_data_1))
ohe_df.head(5)
```

Label Encoding

```python
'''Why we use Label Encoding?
Label Encoding is used to convert categorical text data into numerical data. It assigns a unique integer to each category, which can be useful for algorithms that can interpret the order of categories (like tree-based models). However, it may not be suitable for all algorithms, especially those that assume a linear relationship between features (like linear regression), as it can introduce unintended ordinal relationships between categories.

Difference between One-Hot Encoding and Label Encoding:
1. One-Hot Encoding creates new binary columns for each category, while Label Encoding assigns a unique integer to each category.
2. One-Hot Encoding does not assume any order among categories, while Label Encoding may imply an order (e.g., 0 < 1 < 2).
3. One-Hot Encoding can lead to a large number of features if there are many categories, while Label Encoding keeps the number of features the same as the original categorical column.
4. One-Hot Encoding is generally preferred for nominal categorical variables (no inherent order), while Label Encoding can be used for ordinal categorical variables (with inherent order).

How to choose between One-Hot Encoding and Label Encoding?
- Use One-Hot Encoding when the categorical variable is nominal (no inherent order) and you want to avoid introducing any ordinal relationships between categories.
- Use Label Encoding when the categorical variable is ordinal (with inherent order) and you want to preserve the order of categories.

How to use One-Hot Encoding and Label Encoding in Python?
- For One-Hot Encoding, you can use pandas' get_dummies function or sklearn's OneHotEncoder.
- For Label Encoding, you can use sklearn's LabelEncoder.
'''

# Label Encoding using sklearn
le = LabelEncoder()
le_data = data[en_data_1].apply(le.fit_transform)   # Apply label encoding to the specified columns using the fit_transform method.
le_data.head(10)
```

```python
# Ordinal Encoding using sklearn
oe = OrdinalEncoder()
oe_data = oe.fit_transform(data[en_data_1])  # Fit and transform the data using OrdinalEncoder
oe_df = pd.DataFrame(oe_data, columns=en_data_1)
oe_df.head(10)
```
