```python
import pandas as pd # To import the pandas library to read the dataset

# To read the dataset and to print the first 5 rows of the dataset
dataset = pd.read_excel(
    r"E Commerce Dataset.xlsx", sheet_name="E Comm")
print(dataset.head(5))
```

```python
# To get the information about the dataset and to check the number of missing values in each column
dataset.info()
dataset.isnull().sum()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# To check shape of the data
dataset.shape

#To plot graph of missing values
sns.heatmap(dataset.isnull())
plt.show()
```
